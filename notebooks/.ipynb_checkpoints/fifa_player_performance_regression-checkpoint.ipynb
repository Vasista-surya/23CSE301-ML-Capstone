# Review 1 – Regression Track
## FIFA Player Performance Score Prediction

**Target:** `performance_score`  
**Dataset:** FIFA player-match data (`fifa.csv`)

This notebook follows the Review 1 requirements: dataset audit and EDA, data cleaning, feature engineering, train/test splitting, ten regression algorithms, comparative evaluation, hyperparameter tuning, 5-fold cross-validation for the two best baseline models, and required regression visualisations.

```python
# 1. Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split, KFold, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

sns.set_theme(style='whitegrid')
np.random.seed(42)
```

## 3. Data Cleaning and Feature Engineering

```python
# 3.1 Remove duplicate records and create date-based features
df = df.drop_duplicates().copy()

df['match_date'] = pd.to_datetime(df['match_date'], errors='coerce')
df['match_year'] = df['match_date'].dt.year
df['match_month'] = df['match_date'].dt.month
df.drop(columns=['match_date'], inplace=True)

print('Shape after duplicate removal:', df.shape)
print('Missing values after date conversion:', df.isnull().sum().sum())
```

```python
# 3.2 Create domain-based engineered features
df['goal_contribution_rate'] = (df['goals'] + df['assists']) / (df['minutes_played'].replace(0, np.nan) / 90)
df['work_rate_intensity'] = df['sprint_distance_km'] / (df['distance_covered_km'] + 0.001)
df['passing_efficiency'] = df['successful_passes'] / (df['total_passes'] + 1)

df = df.replace([np.inf, -np.inf], np.nan)

print(df[['goal_contribution_rate', 'work_rate_intensity', 'passing_efficiency']].describe().T)
```

### Feature Engineering Justification
- **Goal contribution rate:** combines goals and assists per 90 minutes so attacking contribution is comparable across playing time.
- **Work-rate intensity:** relates sprint distance to total distance to represent movement intensity.
- **Passing efficiency:** measures successful passes relative to attempted passes and represents passing effectiveness.

```python
# 3.3 Select predictors and remove identifiers/post-outcome variables
# These fields are excluded because they are identifiers or summary scores generated after the match/performance outcome.
drop_columns = [
    'player_id', 'player_name', 'match_id',
    'player_rating', 'tournament_rating',
    'offensive_contribution', 'defensive_contribution',
    'possession_impact', 'pressure_resistance',
    'creativity_score', 'consistency_score', 'clutch_performance_score',
    'total_goals_tournament', 'total_assists_tournament',
    'total_minutes_tournament', 'player_of_match_awards'
]

model_df = df.drop(columns=[c for c in drop_columns if c in df.columns]).copy()

# Use numerical predictors for the ten basic regression models.
model_df = model_df.select_dtypes(include=np.number)

X = model_df.drop(columns=['performance_score'])
y = model_df['performance_score']

print('Predictor shape:', X.shape)
print('Target shape:', y.shape)
```

```python
# 3.4 Train/test split with stratification by target bins
# Bins allow a regression target to be approximately stratified across train and test sets.
y_bins = pd.qcut(y, q=10, labels=False, duplicates='drop')

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y_bins
)

print('Training samples:', X_train.shape[0])
print('Testing samples:', X_test.shape[0])
```

```python
# 3.5 Missing-value treatment using training-set statistics only
train_medians = X_train.median()

X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

# IQR outlier capping: bounds are calculated from training data only.
Q1 = X_train.quantile(0.25)
Q3 = X_train.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

X_train = X_train.clip(lower=lower, upper=upper, axis=1)
X_test = X_test.clip(lower=lower, upper=upper, axis=1)

print('Missing values in training data:', X_train.isnull().sum().sum())
print('Missing values in test data:', X_test.isnull().sum().sum())
```

**Observation:** Missing values are filled using training-set medians, and outlier limits are calculated from the training set before being applied to both splits. This prevents information from the test set from influencing preprocessing.

```python
# 3.6 Standard scaling fitted only on the training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Scaled training shape:', X_train_scaled.shape)
print('Scaled testing shape:', X_test_scaled.shape)
```

## 4. Regression Track – Ten Required Algorithms

```python
# 1. Linear Regression
# 2. Ridge Regression
# 3. Lasso Regression
# 4. ElasticNet Regression
# 5. Polynomial Regression
# 6. Decision Tree Regressor
# 7. Random Forest Regressor
# 8. Gradient Boosting Regressor
# 9. Support Vector Regressor
# 10. K-Nearest Neighbors Regressor

results = {}

def evaluate_model(name, model, X_tr, X_te):
    model.fit(X_tr, y_train)
    predictions = model.predict(X_te)
    return {
        'Algorithm': name,
        'R2': r2_score(y_test, predictions),
        'RMSE': np.sqrt(mean_squared_error(y_test, predictions)),
        'MAE': mean_absolute_error(y_test, predictions),
        'estimator': model,
        'predictions': predictions
    }

results['1. Linear Regression'] = evaluate_model(
    '1. Linear Regression', LinearRegression(), X_train_scaled, X_test_scaled)

results['2. Ridge Regression'] = evaluate_model(
    '2. Ridge Regression', Ridge(alpha=1.0), X_train_scaled, X_test_scaled)

results['3. Lasso Regression'] = evaluate_model(
    '3. Lasso Regression', Lasso(alpha=0.05, max_iter=10000), X_train_scaled, X_test_scaled)

results['4. ElasticNet Regression'] = evaluate_model(
    '4. ElasticNet Regression', ElasticNet(alpha=0.05, l1_ratio=0.5, max_iter=10000), X_train_scaled, X_test_scaled)

# Polynomial Regression uses a manageable subset of strong numerical predictors.
poly_features = ['minutes_played', 'goals', 'assists', 'distance_covered_km', 'expected_goals_xg']
poly_idx = [X.columns.get_loc(c) for c in poly_features if c in X.columns]

poly = PolynomialFeatures(degree=2, include_bias=False)
X_train_poly = poly.fit_transform(X_train_scaled[:, poly_idx])
X_test_poly = poly.transform(X_test_scaled[:, poly_idx])

poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)
poly_pred = poly_model.predict(X_test_poly)

results['5. Polynomial Regression'] = {
    'Algorithm': '5. Polynomial Regression',
    'R2': r2_score(y_test, poly_pred),
    'RMSE': np.sqrt(mean_squared_error(y_test, poly_pred)),
    'MAE': mean_absolute_error(y_test, poly_pred),
    'estimator': poly_model,
    'predictions': poly_pred
}

results['6. Decision Tree Regressor'] = evaluate_model(
    '6. Decision Tree Regressor',
    DecisionTreeRegressor(max_depth=6, random_state=42),
    X_train_scaled, X_test_scaled)

results['7. Random Forest Regressor'] = evaluate_model(
    '7. Random Forest Regressor',
    RandomForestRegressor(n_estimators=50, max_depth=12, random_state=42, n_jobs=1),
    X_train_scaled, X_test_scaled)

results['8. Gradient Boosting Regressor'] = evaluate_model(
    '8. Gradient Boosting Regressor',
    GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=4, random_state=42),
    X_train_scaled, X_test_scaled)

results['9. Support Vector Regressor'] = evaluate_model(
    '9. Support Vector Regressor',
    SVR(C=10, kernel='rbf', epsilon=0.1),
    X_train_scaled, X_test_scaled)

results['10. KNN Regressor'] = evaluate_model(
    '10. KNN Regressor',
    KNeighborsRegressor(n_neighbors=9, weights='distance'),
    X_train_scaled, X_test_scaled)

print('All 10 regression models completed successfully.')
```

```python
# 4.1 Comparative evaluation table
comparison = pd.DataFrame([
    {
        'Algorithm': v['Algorithm'],
        'R2': round(v['R2'], 4),
        'RMSE': round(v['RMSE'], 4),
        'MAE': round(v['MAE'], 4)
    }
    for v in results.values()
]).sort_values('R2', ascending=False).reset_index(drop=True)

display(comparison)
```

**Observation:** The comparison table ranks all ten models using R². RMSE and MAE provide additional measures of prediction error, with lower values indicating smaller errors.

## 5. 5-Fold Cross-Validation for the Two Best Baseline Models

```python
# Select the two best baseline models from the actual comparison table
top_two_names = comparison.head(2)['Algorithm'].tolist()

cv_results = []

for name in top_two_names:
    estimator = results[name]['estimator']
    kf = KFold(n_splits=5, shuffle=True, random_state=42)
    scores = cross_val_score(
        estimator, X_train_scaled, y_train,
        cv=kf, scoring='r2', n_jobs=1
    )
    cv_results.append({
        'Algorithm': name,
        'Mean CV R2': round(scores.mean(), 4),
        'Std CV R2': round(scores.std(), 4)
    })
    print(name, '5-Fold R²:', np.round(scores, 4))

cv_table = pd.DataFrame(cv_results)
display(cv_table)
```

## 6. Hyperparameter Tuning – Two Models

```python
# 6.1 Gradient Boosting Regressor tuning
gb_grid = {
    'n_estimators': [50],
    'learning_rate': [0.05, 0.1],
    'max_depth': [3]
}

gb_search = GridSearchCV(
    GradientBoostingRegressor(random_state=42),
    gb_grid,
    cv=3,
    scoring='r2',
    n_jobs=1
)

gb_search.fit(X_train_scaled, y_train)
gb_pred = gb_search.best_estimator_.predict(X_test_scaled)

gb_base = results['8. Gradient Boosting Regressor']
gb_tuned_r2 = r2_score(y_test, gb_pred)
gb_tuned_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))
gb_tuned_mae = mean_absolute_error(y_test, gb_pred)

print('Best Gradient Boosting Parameters:', gb_search.best_params_)
print(f"Baseline R²: {gb_base['R2']:.4f} | Tuned R²: {gb_tuned_r2:.4f}")
print(f"Baseline RMSE: {gb_base['RMSE']:.4f} | Tuned RMSE: {gb_tuned_rmse:.4f}")
print(f"Baseline MAE: {gb_base['MAE']:.4f} | Tuned MAE: {gb_tuned_mae:.4f}")
```

```python
# 6.2 Random Forest Regressor tuning
rf_grid = {
    'n_estimators': [50],
    'max_depth': [10, 12],
    'min_samples_split': [2, 5]
}

rf_search = GridSearchCV(
    RandomForestRegressor(random_state=42, n_jobs=1),
    rf_grid,
    cv=3,
    scoring='r2',
    n_jobs=1
)

rf_search.fit(X_train_scaled, y_train)
rf_pred = rf_search.best_estimator_.predict(X_test_scaled)

rf_base = results['7. Random Forest Regressor']
rf_tuned_r2 = r2_score(y_test, rf_pred)
rf_tuned_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_tuned_mae = mean_absolute_error(y_test, rf_pred)

print('Best Random Forest Parameters:', rf_search.best_params_)
print(f"Baseline R²: {rf_base['R2']:.4f} | Tuned R²: {rf_tuned_r2:.4f}")
print(f"Baseline RMSE: {rf_base['RMSE']:.4f} | Tuned RMSE: {rf_tuned_rmse:.4f}")
print(f"Baseline MAE: {rf_base['MAE']:.4f} | Tuned MAE: {rf_tuned_mae:.4f}")
```

```python
# 6.3 Compare tuned models
tuned_comparison = pd.DataFrame([
    ['Gradient Boosting Regressor', gb_tuned_r2, gb_tuned_rmse, gb_tuned_mae],
    ['Random Forest Regressor', rf_tuned_r2, rf_tuned_rmse, rf_tuned_mae]
], columns=['Algorithm', 'R2', 'RMSE', 'MAE'])

display(tuned_comparison.sort_values('R2', ascending=False).reset_index(drop=True))
```

## 7. Required Regression Visualisations

```python
# Select the best tuned model using R²
if gb_tuned_r2 >= rf_tuned_r2:
    best_name = 'Gradient Boosting Regressor'
    best_model = gb_search.best_estimator_
    best_predictions = gb_pred
else:
    best_name = 'Random Forest Regressor'
    best_model = rf_search.best_estimator_
    best_predictions = rf_pred

print('Best tuned model:', best_name)
```

```python
# 7.1 Residual plot
residuals = y_test - best_predictions

plt.figure(figsize=(8,5))
plt.scatter(best_predictions, residuals, alpha=0.5)
plt.axhline(0, linestyle='--')
plt.title(f'Residual Plot – {best_name}')
plt.xlabel('Predicted Performance Score')
plt.ylabel('Residual (Actual - Predicted)')
plt.tight_layout()
plt.show()
```

**Observation:** The residual plot is used to check whether errors are centered around zero and whether the error spread changes substantially across predicted values.

```python
# 7.2 Predicted vs actual plot
plt.figure(figsize=(8,5))
plt.scatter(y_test, best_predictions, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], '--', label='Perfect Prediction')
plt.title(f'Predicted vs Actual – {best_name}')
plt.xlabel('Actual Performance Score')
plt.ylabel('Predicted Performance Score')
plt.legend()
plt.tight_layout()
plt.show()
```

**Observation:** Points close to the diagonal reference line indicate predictions close to the actual performance scores. Greater distance from the line represents larger prediction errors.

```python
# 7.3 Feature importance for a tree-based model
feature_model = rf_search.best_estimator_
importance = pd.Series(feature_model.feature_importances_, index=X.columns)
importance = importance.sort_values(ascending=False).head(15)

plt.figure(figsize=(9,6))
importance.sort_values().plot(kind='barh')
plt.title('Top 15 Feature Importances – Random Forest Regressor')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```

**Observation:** The feature-importance plot identifies the predictors that contribute most to the Random Forest's decisions. Importance is model-specific and should not be interpreted as proof of causation.
