```python
# 1. Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

sns.set_theme(style='whitegrid')
np.random.seed(42)
print("comlpleted")
```

```python
# 2.1 Load the raw nutrition dataset
df = pd.read_csv('../datasets/nutrition.csv')

print('Dataset Shape:', df.shape)
print('\nData Types:')
print(df.dtypes.value_counts())
print('\nMissing Values:')
print(df.isnull().sum())
print('\nDuplicate Rows:', df.duplicated().sum())
print('\nOriginal Target Distribution:')
print(df['category'].value_counts())
```

```python
# 2.2 Remove duplicates and create the classification target
df = df.drop_duplicates().copy()

top_categories = df['category'].value_counts().head(10).index
df['category_group'] = df['category'].where(df['category'].isin(top_categories), 'Other')

print('Shape after duplicate removal:', df.shape)
print('\nClassification Target Distribution:')
print(df['category_group'].value_counts())

plt.figure(figsize=(11,5))
df['category_group'].value_counts().plot(kind='bar')
plt.xticks(rotation=45, ha='right')
plt.title('Target Class Distribution')
plt.xlabel('Food Category Group')
plt.ylabel('Number of Records')
plt.tight_layout()
plt.show()
```

```python
# 2.3 Numerical feature distributions
numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols].hist(figsize=(18,18), bins=20)
plt.suptitle('Numerical Feature Distributions', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# 2.4 Categorical distributions
print('Food Category Counts:')
print(df['category'].value_counts())

print('\nFood Name Count:', df['food_name'].nunique())
```

```python
# 2.5 Correlation heatmap
plt.figure(figsize=(15,11))
corr = df[numeric_cols].corr()
plt.imshow(corr, cmap='coolwarm', aspect='auto')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Heatmap – Nutrition Features')
plt.tight_layout()
plt.show()
```

**Observation:** The heatmap shows relationships among calories, macronutrients and micronutrients. Strong correlations can indicate redundant information, while weaker relationships may provide additional classification information.

```python
# 2.6 Feature-target relationship plots
for col in ['calories', 'protein_g', 'fat_g']:
    plt.figure(figsize=(10,5))
    df.boxplot(column=col, by='category_group', rot=45, figsize=(10,5))
    plt.title(f'{col} by Food Category Group')
    plt.suptitle('')
    plt.xlabel('Food Category Group')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# 3.1 Missing values and duplicates
print('Total Missing Values:', df.isnull().sum().sum())
print('Duplicate Rows:', df.duplicated().sum())

# Outlier count using the IQR rule
outlier_counts = {}
for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    outlier_counts[col] = ((df[col] < q1 - 1.5*iqr) | (df[col] > q3 + 1.5*iqr)).sum()

print('\nOutlier Counts:')
print(pd.Series(outlier_counts).sort_values(ascending=False))
```

```python
# 4.1 Feature engineering
df['total_macros_g'] = (
    df['protein_g'] + df['carbohydrates_g'] + df['fat_g']
)

df['fat_carb_ratio'] = df['fat_g'] / (df['carbohydrates_g'] + 1)
df['protein_carb_ratio'] = df['protein_g'] / (df['carbohydrates_g'] + 1)

print(df[['total_macros_g', 'fat_carb_ratio', 'protein_carb_ratio']].describe().T)
```

```python
# 4.2 Prepare features and target
X = df.drop(columns=['category', 'category_group', 'food_name']).select_dtypes(include=np.number)
y = df['category_group']

print('Feature Matrix Shape:', X.shape)
print('Target Shape:', y.shape)
```

```python
# 4.3 Stratified 80:20 train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print('Training samples:', X_train.shape[0])
print('Testing samples:', X_test.shape[0])
```

```python
# 4.4 Missing values and outlier treatment using training data only
train_medians = X_train.median()
X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

Q1 = X_train.quantile(0.25)
Q3 = X_train.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

X_train = X_train.clip(lower=lower, upper=upper, axis=1)
X_test = X_test.clip(lower=lower, upper=upper, axis=1)

print('Training missing values:', X_train.isnull().sum().sum())
print('Testing missing values:', X_test.isnull().sum().sum())
```

```python
# 4.5 Standard scaling fitted only on the training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('Scaled training shape:', X_train_scaled.shape)
print('Scaled testing shape:', X_test_scaled.shape)
```

```python
# Classification Models 1–2

# 1. Logistic Regression
log_model = LogisticRegression(max_iter=2000, random_state=42)
log_model.fit(X_train_scaled, y_train)
log_pred = log_model.predict(X_test_scaled)

log_accuracy = accuracy_score(y_test, log_pred)
log_f1 = f1_score(y_test, log_pred, average='weighted')

print('1. Logistic Regression')
print(f'Accuracy   : {log_accuracy:.4f}')
print(f'Weighted F1: {log_f1:.4f}')

# 2. K-Nearest Neighbors
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled, y_train)
knn_pred = knn_model.predict(X_test_scaled)

knn_accuracy = accuracy_score(y_test, knn_pred)
knn_f1 = f1_score(y_test, knn_pred, average='weighted')

print('\n2. K-Nearest Neighbors')
print(f'Accuracy   : {knn_accuracy:.4f}')
print(f'Weighted F1: {knn_f1:.4f}')
```

```python
# Classification Models 3–4

# 3. Gaussian Naive Bayes
gnb_model = GaussianNB()
gnb_model.fit(X_train_scaled, y_train)
gnb_pred = gnb_model.predict(X_test_scaled)

gnb_accuracy = accuracy_score(y_test, gnb_pred)
gnb_f1 = f1_score(y_test, gnb_pred, average='weighted')

print('3. Gaussian Naive Bayes')
print(f'Accuracy   : {gnb_accuracy:.4f}')
print(f'Weighted F1: {gnb_f1:.4f}')

# 4. Decision Tree Classifier
dt_model = DecisionTreeClassifier(max_depth=6, random_state=42)
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)

dt_accuracy = accuracy_score(y_test, dt_pred)
dt_f1 = f1_score(y_test, dt_pred, average='weighted')

print('\n4. Decision Tree Classifier')
print(f'Accuracy   : {dt_accuracy:.4f}')
print(f'Weighted F1: {dt_f1:.4f}')
```

```python
# Classification Model 5

# 5. Support Vector Classifier
svc_model = SVC(C=1.0, kernel='rbf', random_state=42)
svc_model.fit(X_train_scaled, y_train)
svc_pred = svc_model.predict(X_test_scaled)

svc_accuracy = accuracy_score(y_test, svc_pred)
svc_f1 = f1_score(y_test, svc_pred, average='weighted')

print('5. Support Vector Classifier')
print(f'Accuracy   : {svc_accuracy:.4f}')
print(f'Weighted F1: {svc_f1:.4f}')
```

```python
# Final Classification Evaluation
predictions = {
    '1. Logistic Regression': log_pred,
    '2. KNN': knn_pred,
    '3. Gaussian Naive Bayes': gnb_pred,
    '4. Decision Tree': dt_pred,
    '5. SVC': svc_pred
}

comparison = []

for name, pred in predictions.items():
    accuracy = accuracy_score(y_test, pred)
    weighted_f1 = f1_score(y_test, pred, average='weighted')

    comparison.append({
        'Algorithm': name,
        'Accuracy': round(accuracy, 4),
        'Weighted F1': round(weighted_f1, 4)
    })

comparison_table = pd.DataFrame(comparison)

print('Classification Model Comparison:')
display(comparison_table)

# Confusion matrices for all five classifiers
class_labels = sorted(y_test.unique())

for name, pred in predictions.items():
    cm = confusion_matrix(y_test, pred, labels=class_labels)

    plt.figure(figsize=(9, 7))
    plt.imshow(cm, interpolation='nearest', cmap='Blues')
    plt.colorbar()
    plt.xticks(
        range(len(class_labels)),
        class_labels,
        rotation=45,
        ha='right'
    )
    plt.yticks(range(len(class_labels)), class_labels)

    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(
                j, i, cm[i, j],
                ha='center',
                va='center',
                fontsize=7
            )

    plt.title(f'Confusion Matrix – {name}')
    plt.xlabel('Predicted Class')
    plt.ylabel('Actual Class')
    plt.tight_layout()
    plt.show()

# Sorted comparison for interpretation
classification_comparison = comparison_table.sort_values(
    'Weighted F1',
    ascending=False
).reset_index(drop=True)

print('Sorted Classification Comparison:')
display(classification_comparison)
```

```python
# Visualise the required Decision Tree model
plt.figure(figsize=(20,10))
plot_tree(
    dt_model,
    feature_names=X.columns,
    class_names=sorted(y.unique()),
    filled=True,
    max_depth=3,
    fontsize=8
)
plt.title('Decision Tree Classifier – First Three Levels')
plt.tight_layout()
plt.show()
```

# Nutrition Dataset — Classification Review 2 (Part B Only)

**Scope:** This notebook contains only the five Part-B classifiers: **Random Forest, AdaBoost, Gradient Boosting, Bagging, and MLP Classifier**. Each model has its own independent cell that trains the model and reports classification metrics, a per-class classification report, and a confusion matrix.

**Evaluation protocol:** stratified 80:20 train/test split with `random_state=42`. Imputation, IQR outlier capping, and scaling (for MLP) are fitted using training data only. The target follows the existing nutrition notebook: the ten most frequent food categories remain separate and all others are grouped as `Other`.


## 1. Import libraries

```python
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
RANDOM_STATE = 42

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier, BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score
)

print("Imports completed.")
```

## 2. Load and prepare the Nutrition dataset

The loader checks common relative locations so the notebook can run from the project root or from a notebooks folder. If needed, change `DATA_PATH` to the location of your `nutrition.csv` file.

```python
possible_paths = [
    "../datasets/nutrition.csv",
    "datasets/nutrition.csv",
    "../nutrition.csv",
    "nutrition.csv",
    "/mnt/data/nutrition.csv",
]

DATA_PATH = next((path for path in possible_paths if os.path.exists(path)), None)
if DATA_PATH is None:
    raise FileNotFoundError(
        "Could not find nutrition.csv. Place it in ../datasets/, datasets/, "
        "the current folder, or update DATA_PATH to the correct file path."
    )

df = pd.read_csv(DATA_PATH).drop_duplicates().copy()
print("Loaded:", DATA_PATH)
print("Dataset shape after removing duplicates:", df.shape)
display(df.head())
```

## 3. Create target, engineer features, and split data

```python
# Keep the same target definition used by the original Nutrition classification notebook.
if "category" not in df.columns:
    raise KeyError("The dataset must contain a 'category' column.")
if "food_name" not in df.columns:
    print("Note: 'food_name' column not found; it will simply be excluded if present.")

top_categories = df["category"].value_counts().head(10).index
df["category_group"] = df["category"].where(df["category"].isin(top_categories), "Other")

# Feature engineering consistent with the existing notebook.
for col in ["protein_g", "carbohydrates_g", "fat_g"]:
    if col not in df.columns:
        raise KeyError(f"Required feature column '{col}' is missing.")
df["total_macros_g"] = df["protein_g"] + df["carbohydrates_g"] + df["fat_g"]
df["fat_carb_ratio"] = df["fat_g"] / (df["carbohydrates_g"] + 1)
df["protein_carb_ratio"] = df["protein_g"] / (df["carbohydrates_g"] + 1)

drop_cols = [c for c in ["category", "category_group", "food_name", "name"] if c in df.columns]
X = df.drop(columns=drop_cols).select_dtypes(include=np.number)
y = df["category_group"]

if X.shape[1] == 0:
    raise ValueError("No numeric feature columns were found after target/text columns were excluded.")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

# Fit preprocessing statistics on training data only.
train_medians = X_train.median()
X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

Q1 = X_train.quantile(0.25)
Q3 = X_train.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
X_train = X_train.clip(lower=lower, upper=upper, axis=1)
X_test = X_test.clip(lower=lower, upper=upper, axis=1)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

class_labels = sorted(y.unique())
print(f"Features: {X.shape[1]}")
print(f"Training samples: {len(X_train)} | Test samples: {len(X_test)}")
print(f"Classes ({len(class_labels)}): {class_labels}")
print("Train/test split: stratified 80:20")
```

## 4. Evaluation helper

Every Part-B model cell below calls this helper. It reports **accuracy, weighted precision, weighted recall, weighted F1-score, ROC-AUC (OvR when available), a per-class precision/recall/F1/support report, and a labelled confusion matrix**. Confusion-matrix rows are actual labels and columns are predicted labels.

```python
part_b_results = {}

def evaluate_classifier(name, model, X_eval, y_true):
    y_pred = model.predict(X_eval)

    metrics = {
        "Algorithm": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision (weighted)": precision_score(y_true, y_pred, average="weighted", zero_division=0),
        "Recall (weighted)": recall_score(y_true, y_pred, average="weighted", zero_division=0),
        "F1-score (weighted)": f1_score(y_true, y_pred, average="weighted", zero_division=0),
    }

    # Multiclass ROC-AUC can be unavailable for a model/dataset; report clearly rather than failing.
    try:
        if hasattr(model, "predict_proba"):
            y_score = model.predict_proba(X_eval)
            metrics["ROC-AUC (OvR, weighted)"] = roc_auc_score(
                y_true, y_score, labels=model.classes_, multi_class="ovr", average="weighted"
            )
        elif hasattr(model, "decision_function"):
            y_score = model.decision_function(X_eval)
            metrics["ROC-AUC (OvR, weighted)"] = roc_auc_score(
                y_true, y_score, labels=model.classes_, multi_class="ovr", average="weighted"
            )
        else:
            metrics["ROC-AUC (OvR, weighted)"] = np.nan
    except Exception as exc:
        metrics["ROC-AUC (OvR, weighted)"] = np.nan
        print("ROC-AUC unavailable for this evaluation:", str(exc))

    print(f"\n{'='*72}\n{name}\n{'='*72}")
    print("Summary metrics:")
    display(pd.DataFrame([metrics]).round(4))

    print("\nPer-class classification report:")
    print(classification_report(y_true, y_pred, labels=class_labels, zero_division=0))

    cm = confusion_matrix(y_true, y_pred, labels=class_labels)
    cm_df = pd.DataFrame(cm, index=class_labels, columns=class_labels)
    print("Confusion matrix (rows = actual, columns = predicted):")
    display(cm_df)

    fig_width = max(10, min(18, 1.1 * len(class_labels)))
    plt.figure(figsize=(fig_width, max(7, 0.75 * len(class_labels))))
    sns.heatmap(cm_df, annot=True, fmt="d", cmap="Blues", cbar=True, square=False)
    plt.title(f"Confusion Matrix — {name}")
    plt.xlabel("Predicted class")
    plt.ylabel("Actual class")
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    plt.close()

    part_b_results[name] = {**metrics, "model": model, "predictions": y_pred, "confusion_matrix": cm}
    return metrics
```

## 5. Random Forest

Train this model independently and inspect its metrics and confusion matrix below.

```python
# Part B — Model 1: Random Forest
random_forest_model = RandomForestClassifier(
    n_estimators=100, max_depth=None, random_state=RANDOM_STATE, n_jobs=-1
)
random_forest_model.fit(X_train, y_train)
rf_metrics = evaluate_classifier("Random Forest", random_forest_model, X_test, y_test)
```

## 5. AdaBoost

Train this model independently and inspect its metrics and confusion matrix below.

```python
# Part B — Model 2: AdaBoost
adaboost_model = AdaBoostClassifier(
    n_estimators=100, learning_rate=0.5, random_state=RANDOM_STATE
)
adaboost_model.fit(X_train, y_train)
ada_metrics = evaluate_classifier("AdaBoost", adaboost_model, X_test, y_test)
```

## 5. Gradient Boosting

Train this model independently and inspect its metrics and confusion matrix below.

```python
# Part B — Model 3: Gradient Boosting
gradient_boosting_model = GradientBoostingClassifier(
    n_estimators=100, learning_rate=0.1, max_depth=3, random_state=RANDOM_STATE
)
gradient_boosting_model.fit(X_train, y_train)
gb_metrics = evaluate_classifier("Gradient Boosting", gradient_boosting_model, X_test, y_test)
```

## 5. Bagging

Train this model independently and inspect its metrics and confusion matrix below.

```python
# Part B — Model 4: Bagging
# Handle both newer and older scikit-learn parameter names.
try:
    bagging_model = BaggingClassifier(
        estimator=DecisionTreeClassifier(random_state=RANDOM_STATE),
        n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1
    )
except TypeError:
    bagging_model = BaggingClassifier(
        base_estimator=DecisionTreeClassifier(random_state=RANDOM_STATE),
        n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1
    )

bagging_model.fit(X_train, y_train)
bag_metrics = evaluate_classifier("Bagging", bagging_model, X_test, y_test)
```

## 5. MLP Classifier

Train this model independently and inspect its metrics and confusion matrix below.

```python
# Part B — Model 5: Multi-Layer Perceptron (MLP) Classifier
mlp_model = MLPClassifier(
    hidden_layer_sizes=(64, 32), activation="relu", solver="adam",
    max_iter=400, early_stopping=True, random_state=RANDOM_STATE
)
mlp_model.fit(X_train_scaled, y_train)
mlp_metrics = evaluate_classifier("MLP Classifier", mlp_model, X_test_scaled, y_test)
```

## 6. Compare the five Part-B classifiers

```python
comparison_rows = []
for name, result in part_b_results.items():
    comparison_rows.append({
        key: value for key, value in result.items()
        if key in ["Algorithm", "Accuracy", "Precision (weighted)", "Recall (weighted)",
                   "F1-score (weighted)", "ROC-AUC (OvR, weighted)"]
    })

part_b_comparison = pd.DataFrame(comparison_rows).sort_values(
    "F1-score (weighted)", ascending=False
).reset_index(drop=True)
display(part_b_comparison.round(4))

print("Best model by weighted F1-score:", part_b_comparison.iloc[0]["Algorithm"])
```

## 7. Interpretation guide

- **Confusion matrix:** diagonal values are correct predictions; off-diagonal values show which food categories the model confuses.
- **Precision:** among items predicted as a class, how many truly belong to that class.
- **Recall:** among items actually in a class, how many the model identifies correctly.
- **F1-score:** harmonic mean of precision and recall.
- **Weighted averages:** metrics are averaged across classes while accounting for each class's support.
- **ROC-AUC (OvR):** measures class-separation ability using a one-vs-rest approach; it may be unavailable in edge cases.

Use the comparison table and the per-model confusion matrices together. A high overall score can hide poor performance on less frequent classes, so inspect the per-class report as well.
